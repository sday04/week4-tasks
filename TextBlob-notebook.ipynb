<h2 id="sec1"> Create a TextBlob </h2>

```python
from textblob import TextBlob
```

```python
wiki = TextBlob("Python is a high-level, general-purpose programming language.")
```

```python
wiki.tags
```

<h2 id="sec3"> Noun Phrase Extraction </h2>

<h2 id="sec4"> Sentiment Analysis </h2>

```python
testimonial = TextBlob("Textblob is amazingly simple to use. What great fun!")
print(testimonial.sentiment)
print(testimonial.sentiment.polarity)
```

<h2 id="sec5"> Tokenization </h2>

breaking down TextBlobs into words or sentences

```python
from textblob import TextBlob

zen = TextBlob("Beautiful is better than ugly. "
               "Explicit is better than implicit. "
               "Simple is better than complex.")
```

```python
zen.words
```

```python
zen.sentences
```

```python
for sentence in zen.sentences:
    print(sentence.sentiment)
```

<h2 id="sec6"> Words Inflection and Lemmatization </h2>

```python
sentence = TextBlob("Use 4 spaces per indentation level.")
sentence.words
```

```python
sentence.words[2].singularize()
```

```python
sentence.words[-1].pluralize()
```

```python
# using lemmatize method to lemmatize words
from textblob import Word
w = Word("octopi")
w.lemmatize()
```

```python
w = Word("went")
w.lemmatize("v") # verb
```

<h2 id="sec7"> WordNet Integration </h2>

```python
from textblob.wordnet import VERB
word = Word("octopus")
word.synsets
```

```python
Word("hack").get_synsets(pos=VERB)
```

```python
# Creating Synsets
from textblob.wordnet import Synset
octopus = Synset("octopus.n.02")
shrimp = Synset("shrimp.n.03")
octopus.path_similarity(shrimp) # how similar two words are
```

<h2 id="sec10"> Get Word and Noun Phrase Frequencies </h2>

```python
monty = TextBlob("We are no longer the Knights who say Ni. "
                 "We are now the Knights who say Ekki ekki ekki PTANG.")
monty.word_counts['ekki']
```

```python
# can also do
monty.words.count('ekki')
```

```python
# can make case sensitive
monty.words.count('ekki', case_sensitive=True)
```

```python
# can be used with noun phrases
wiki.noun_phrases.count('python')
```

<h2 id="sec11"> Parsing </h2>

```python
b = TextBlob("And now for something completely different.")
print(b.parse())
```

<h2 id="sec12"> TextBlobs Are Like Python Strings </h2>

```python
# Can use python string operations on textblobs
zen[0:19]
```

```python
zen.upper()
```

```python
zen.find("Simple")
```

```python
apple_blob = TextBlob("apples")
banana_blob = TextBlob("bananas")
print(apple_blob < banana_blob)
print(apple_blob == "apples")
```

```python
apple_blob + " and " + banana_blob
```

```python
"{0} and {1}".format(apple_blob, banana_blob)
```

<h2 id="sec13"> n-grams </h2>
Returns list of tuples of n successive words

```python
blob = TextBlob("Now is better than never.")
blob.ngrams(n=3)
```

<h2 id="sec14"> Get Start and End Indices of Sentences </h2>

```python
for s in zen.sentences:
    print(s)
    print("---- Starts at index {}, Ends at index {}".format(s.start, s.end))
```
